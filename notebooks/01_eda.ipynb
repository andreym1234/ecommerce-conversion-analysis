```python
path = "../data/raw/2019-Oct.csv"

total_rows = 0
missing_count = 0

for chunk in pd.read_csv(path, usecols=["category_code"], chunksize=100_000):
    total_rows += len(chunk)
    missing_count += chunk["category_code"].isna().sum()

missing_rate = missing_count / total_rows * 100

print(f"Всего строк: {total_rows}")
print(f"Пропусков: {missing_count}")
print(f"Доля пропусков: {missing_rate:.2f}%")
```

```python
path = "../data/raw/2019-Oct.csv"

total_rows = 0
missing_count = 0

for chunk in pd.read_csv(path, usecols=["user_session"], chunksize=100_000):
    total_rows += len(chunk)
    missing_count += chunk["user_session"].isna().sum()

missing_rate = missing_count / total_rows * 100

print(f"Всего строк: {total_rows}")
print(f"Пропусков: {missing_count}")
print(f"Доля пропусков: {missing_rate:.2f}%")
```

```python
event_counts = pd.Series(dtype="int64")

for chunk in pd.read_csv(path, usecols=["event_type"], chunksize=100_000):
    counts = chunk["event_type"].value_counts()
    event_counts = event_counts.add(counts, fill_value=0)

event_counts = event_counts.astype("int64")
print(event_counts.sort_values(ascending=False))
```

```python
path = "../data/raw/2019-Oct.csv"

total_rows = 0
missing_counts = pd.Series(dtype="int64")

for chunk in pd.read_csv(path, chunksize=100_000):
    total_rows += len(chunk)
    missing_counts = missing_counts.add(chunk.isna().sum(), fill_value=0)

missing_stats = pd.DataFrame({
    "missing_count": missing_counts.astype("int64"),
    "missing_percent": (missing_counts / total_rows * 100).round(2)
})

missing_stats = missing_stats.sort_values(
    by="missing_percent",
    ascending=False
)

display(missing_stats)
```

```python
min_date = None
max_date = None

invalid_dates = 0
zero_prices = 0
negative_prices = 0

for chunk in pd.read_csv(
    path,
    usecols=["event_time", "price"],
    chunksize=100_000
):
    dates = pd.to_datetime(
        chunk["event_time"],
        utc=True,
        errors="coerce"
    )

    invalid_dates += dates.isna().sum()
    zero_prices += (chunk["price"] == 0).sum()
    negative_prices += (chunk["price"] < 0).sum()

    chunk_min = dates.min()
    chunk_max = dates.max()

    if pd.notna(chunk_min):
        min_date = chunk_min if min_date is None else min(min_date, chunk_min)

    if pd.notna(chunk_max):
        max_date = chunk_max if max_date is None else max(max_date, chunk_max)

print(f"Начало периода: {min_date}")
print(f"Конец периода: {max_date}")
print(f"Некорректные даты: {invalid_dates}")
print(f"Нулевые цены: {zero_prices}")
print(f"Отрицательные цены: {negative_prices}")
```

```python
purchase_count = 0
zero_purchase = 0
negative_purchase = 0

for chunk in pd.read_csv(
    path,
    usecols=["event_type", "price"],
    chunksize=100_000
):
    purchases = chunk[chunk["event_type"] == "purchase"]

    purchase_count += len(purchases)
    zero_purchase += (purchases["price"] == 0).sum()
    negative_purchase += (purchases["price"] < 0).sum()

print(f"Всего покупок (событий): {purchase_count}")
print(f"Покупок с нулевой ценой: {zero_purchase}")
print(f"Покупок с отрицательной ценой: {negative_purchase}")
```

```python
users = set()
products = set()
sessions = set()

for chunk in pd.read_csv(
    path,
    usecols=["user_id", "product_id", "user_session"],
    chunksize=100_000
):
    users.update(chunk["user_id"].dropna().unique())
    products.update(chunk["product_id"].dropna().unique())
    sessions.update(chunk["user_session"].dropna().unique())

print(f"Уникальных пользователей: {len(users):,}")
print(f"Уникальных товаров: {len(products):,}")
print(f"Уникальных ID сессий: {len(sessions):,}")
```

```python
session_to_user = {}
conflicting_sessions = set()

for chunk in pd.read_csv(
    path,
    usecols=["user_id", "user_session"],
    chunksize=100_000
):
    pairs = chunk.dropna().drop_duplicates()

    for user, session in pairs.itertuples(index=False, name=None):
        if session in session_to_user:
            if session_to_user[session] != user:
                conflicting_sessions.add(session)
        else:
            session_to_user[session] = user

print(f"Всего ID сессий: {len(session_to_user):,}")
print(f"ID сессий у разных пользователей: {len(conflicting_sessions):,}")
print(f"Доля конфликтных ID: {len(conflicting_sessions) / len(session_to_user) * 100:.4f}%")
```
